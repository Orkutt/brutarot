# Observe tarot dataset
## 1. Import modules   

* `pandas` library and `json_normalize` function from `pandas.io.json` to read .json document into a tabular-like dataset;
* `PIL` (Python Imaging Library, or Pillow) library to display the images;
* `seaborn` and `matplotlib` to create plots with images.

````python
import pandas as pd
from pandas import json_normalize
import PIL                      #read images
import matplotlib.pyplot as plt #show images
import seaborn as sns
sns.set_theme(style="whitegrid")
````

## 2. Import data

File `tarot-images.json` includes 78 cards.   
Every card includes `name`, `number`, `arcana` (Major or Minor), `suit`, `img` (reference to the picture), `fortune telling` (3 phrases), four `keywords`, `meanings` (light and shadow), `Questions to Ask`. Other options depend on a card type.

````python
cardsImgs = pd.read_json("tarot-images.json", orient='records')
cardsImgs = json_normalize(cardsImgs['cards'])
cardsImgs
````

🔮 we have 78 entries, where 22 cards do not have a suit (Major Arcana).   
Most of the cells in dataframe contains inside-lists, f.e. column `keywords` consists of descriptive words about the card.

````python
cardsImgs.info()
````

## 3. Fast look - cards by Arcana


* **Major Arcana** represents archetypes: crucial aspects in every life and situation.  
* **Minor Arcana** represents everyday details:   
```
    - WANDS (creativity, desire) 
    - CUPS (emotion, relationship)
    - SWORDS (logic, reason) 
    - PENTACLES (finances)  
```

All cards have name, number, arcana, suit, corresponding image, fortune telling, questions to ask (list) and meanings (light and shadow) - **78**  
All cards, except Page, Knight, Queen, King of every suit have Numerology series - **62**  
Only Minor Arcana cards have Affirmation - **56**  
Major Arcana, Page, Knight, Queen, King of every suit has Astrology series - **40**  
Every Minor Arcana card except Page, Knight, Queen, King have Elemental series - **38**  
Only Major Arcana cards have Archetype and Hebrew Alphabet - **22**  


🔮 Bar plot of cards by Arcana: 56 minor and 22 major  
14 cards of every suit: Cups, Swords, Wands and Pentacles (Minor Arcana)  
22 cards of Major Arcana: Trumps

````python
ax = sns.countplot(x="arcana", hue="suit", data=cardsImgs)
ax.legend(loc='center right', bbox_to_anchor=(1.25, 0.5), ncol=1)
for p in ax.patches:
    ax.annotate('{}'.format(p.get_height()), (p.get_x(), p.get_height()+0.5))
ax
````

# Get reading
🔮 Get 3 random cards: `Past, Present, Future`. This simple spread is built to give a general comprehension.

````python
import datetime
print("Your reading: Past, Present, Future on", datetime.datetime.now().date())

cards3 = cardsImgs.sample(n = 3).reset_index(drop=True)

plt.figure(figsize=(10,10))
c = 0
for i in cards3['img']:
    img = PIL.Image.open(f'cards/{i}')
    plt.subplot(1,3,c+1)
    plt.imshow(img)
    plt.axis('off')
    c+=1

plt.show()
````

````python
t = ['PAST', 'PRESENT', 'FUTURE']

for index, row in cards3.iterrows():
    print('\x1b[31m{}\x1B[37m: {} \n   {}, {}'.format(t[index], row['name'], row['arcana'], row['suit']))
    for r in row['fortune_telling']:
        print("\x1B[37m\t --->\x1b[31m", r.upper())
    for r in row['Questions to Ask']:
        print("\x1B[37m\t    ?", r)
    print(10*'-')
````
